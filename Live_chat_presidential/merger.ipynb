```python
import json
import pandas as pd
```

```python
# Load both JSON files
file1_path = './wfaa_DK.json'  # Path to first file
file2_path = './wfaa_AB.json'  # Path to second file

# Load the JSON data from both files

with open(file1_path, 'r') as file1, open(file2_path, 'r') as file2:
    data1 = json.load(file1)
    data2 = json.load(file2)

# Convert to DataFrames
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

```python
df1['timestamp'] = pd.to_datetime(df1['timestamp'])
df2['timestamp'] = pd.to_datetime(df2['timestamp'])

# sort the dataframes by timestamp
df1 = df1.sort_values('timestamp')
df2 = df2.sort_values('timestamp')

df1.head()
```

```python
# define merged start and end time.
start_timestamp = min(df1["timestamp"].min(), df2["timestamp"].min())
end_timestamp = max(df1["timestamp"].max(), df2["timestamp"].max())

# pick data from start - from min(df1["timestamp"].min(), df2["timestamp"].min()) to max(df1["timestamp"].min(), df2["timestamp"].min())
start_range = (start_timestamp, max(df1["timestamp"].min(), df2["timestamp"].min()))
# pick data from end
end_range = (min(df1["timestamp"].max(), df2["timestamp"].max()), end_timestamp)
```

```python
start_range
```

```python
def merge_start(range, df1, df2):
    start_at = range[0]

    if df1["timestamp"][0] == start_at:
        # pick elements from df1 for given range
        df1 = df1[df1["timestamp"] <= range[1]]
        return df1
    else:
        # pick elements from df2 for given range
        df2 = df2[df2["timestamp"] <= range[1]]
        return df2

def merge_end(range, df1, df2):
    end_at = range[1]
    
    # compare last timestamp of df1
    if df1["timestamp"].iloc[-1] == end_at:
        # pick elements from df1 for given range
        df1 = df1[df1["timestamp"] >= range[0]]
        return df1
    else:
        # pick elements from df2 for given range
        df2 = df2[df2["timestamp"] >= range[0]]
        return df2
    
# mid_range = (start_range[1], end_range[0])

# merge from one after end of merge_start to one before start of merge_end
df_mid = df1[(df1["timestamp"] > start_range[1]) & (df1["timestamp"] < end_range[0])]

```
